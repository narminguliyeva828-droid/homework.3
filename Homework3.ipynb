Task1 Grading System

```python
mark=int(input("enter your exam mark: "))
if mark>90:
    print("Your grade is A")
elif mark>74 and mark<90:
    print("Your grade is B")
elif mark>49 and mark<75:
    print("Your grade is C")
else:
    print("Your grade is F")
```

Task2 Multiplication Table for loop

```python
n=int(input("enter your number: "))
print(f" Multiplication table of {n} is: ")
for i in range(1, 11):
    print(f"{n} x {i} = {n*i}")
```

Task3 Password Retry System while loop 

```python
correctpass=20081204
passw=input("enter your password: ")
attepts=0
while passw!=correctpass and attepts<3:
    print("Incorrect password, try again")
    attepts+=1
    passw=input("enter your password: ")
if attepts==3:
    print("many attempts, try again later")
```
